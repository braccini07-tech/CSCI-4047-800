```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("./Income Data.csv")

```

3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
def fns(column): 
    min = np.min(df[column])
    Q1 = np.percentile(df[column], 25)
    median = np.median(df[column])
    Q3 = np.percentile(df[column], 75)
    max = np.max(df[column])
    mean = np.mean(df[column])
    print (f"{column}---------------")
    print (f"min: {min}")
    print (f" Q1: {Q1}")
    print (f" median: {median}")
    print (f" Q3: {Q3}")
    print (f" max: {max}")
    print (f" mean: {mean}")

fns("income")
fns("education")
fns("prestige")
```

4. Plot a *histogram* for **prestige**
5. Plot a *histogram* for **education**

```python
plt.hist(df['prestige'], color= 'blue', alpha = 0.5)
plt.show()
plt.hist(df['education'], color = 'blue', alpha = 0.5)
plt.show()
```

6. Plot a *scatter plot* for **income** and **education**
7. Plot a *scatter plot* for **education** and **prestige**

```python
plt.scatter(x=df ['education'], y= df ['income'])
plt.title("Income compared to Educaiton")
plt.xlabel('Education')
plt.ylabel('Income')
plt.show()

plt.scatter(x = df ['prestige'], y = df ['education'])
plt.title("Education compared to Prestige")
plt.xlabel('Prestige')
plt.ylabel('Education')
plt.show()
```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python
standard_dev = df ['income'].std()
print(f"Standard Deviation: {standard_dev: .2f}")

variance = standard_dev **2
print(f"Variance: {variance: .2f}")
```
