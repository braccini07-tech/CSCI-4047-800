```python
# import pandas (and numpy, scipy.stats, statsmodels.api as you need them)
import pandas as pd
import numpy as np
import scipy.stats as stats
import statsmodels.api as sm
# load customers.csv into a DataFrame
df = pd.read_csv('./customers.csv')
# take a first look (head, info)
df.head()
df.info()
```

## Part 0: Know your data, then clean it

This data is dirty on purpose. Find and fix **three** problems before you analyze anything.

```python
# HINT 1: a numeric column has missing values -> decide how to fill them (mean? median? why?)
df["income"]= df["income"].fillna(84.7)
# HINT 2: one category label is written in the wrong case -> standardize it
region_map = {
    "east" : "East"
}
df['region']= df['region'].replace(region_map)
# HINT 3: one value is impossible for its column -> treat it as missing, then fill
#print(df.describe())
#df['income'] = df["income"].replace(-50, np.nan)
df["income"] = df["income"].fillna(df["income"].median())
# tip: check .isna().sum() and the unique values of the category column
# print(df.isna().sum())
# income has 2 blank values
print(df.isna().sum())

  
```

**Explain each issue:** what you found and how you handled it.

*Your answer:* The first issue was two missing values in the income column. The missing values were filled using the median income of 84.7. The median was used because income can contain unusually high or low values that could affect the mean.

The second issue was a capitalization error in the region column. One value was written as "east" instead of "East". This was corrected by replacing "east" with "East" so the category labels were consistent.

The third issue was an impossible income value of -50. Since a negative income value was not valid for this dataset, it was first replaced with NaN and then filled using the median income.

### Q1. Do customers who earn more also spend more?

```python
# data types: income = continuous, spending = continuous 
# tool: pearson correlation
# hint: two continuous variables moving together
r, p = stats.pearsonr(df["income"], df["spending"])

print("Correlation:", r)
print("P-value:", p)
```

*Your answer:* Yes, customers who earn more generally spend more, and the relationship is statistically significant.


### Q2. Is a customer's region related to whether they subscribed?

```python
# data types: region = ?, subscribed = ?
# tool: Chi-square test
# hint: two categorical variables -> build a table with pd.crosstab, then test it
table = pd.crosstab(df["region"], df["subscribed"])
chi2, p, dof, expected = stats.chi2_contingency(table)

print("Chi-square:", chi2)
print("P-value:", p)
```

*Your answer:* Yes, a customer's region is significantly related to whether they subscribed.
